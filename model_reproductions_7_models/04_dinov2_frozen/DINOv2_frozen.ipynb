# DINOv2 ViT-S/14 — frozen pretrained feature extractor

The model is defined locally in `dinov2_model.py`; it is an explicit ViT-S/14 architecture with DINOv2 pretrained weights.


```python
import sys
from pathlib import Path

MODEL_FOLDER = (
    Path.home() / "Desktop/Paper replication/model_reproductions_7_models"
    / "04_dinov2_frozen"
)
sys.path.insert(0, str(MODEL_FOLDER))

from dinov2_model import build_dinov2_small, load_dinov2_pretrained_backbone
from vit_training_helpers import configure_parameters, set_seed, train_experiment
```

```python
set_seed(42)
model = build_dinov2_small()
checkpoint = MODEL_FOLDER/'pretrained/dinov2_vits14_pretrain.pth'
load_dinov2_pretrained_backbone(model, checkpoint)
print(model)
```

```python
configure_parameters(model, phase='frozen', last_stage=None)
print('Trainable:', sum(p.numel() for p in model.parameters() if p.requires_grad))
print('Total:', sum(p.numel() for p in model.parameters()))
```

```python
train_experiment(
    model=model, phase='frozen', last_stage=model.blocks[-1],
    output_dir=MODEL_FOLDER/'notebook_results',
    epochs=20, learning_rate=1e-4, batch_size=16, center_crop=True,
)
```
